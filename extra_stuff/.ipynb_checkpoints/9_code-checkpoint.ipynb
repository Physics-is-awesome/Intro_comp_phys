```python
import numpy as np
import matplotlib.pyplot as plt


def f(t, state):
    x, y, dx, dy = state

    f0 = dx
    f1 = dy

    f2 = sgn * (1/m) * 2*y**2 * x * (1 - x**2) * np.exp(-(x**2 + y**2))
    f3 = sgn * (1/m) * 2*x**2 * y * (1 - y**2) * np.exp(-(x**2 + y**2))

    return np.array([f0, f1, f2, f3])





arrays = {}

N = 1000
sgn = +1
m = 0.5
dt = 0.01

for i in range(10):

    # One random initial condition per run
    x0 = np.random.randn()
    y0 = np.random.randn()
    dx0 = np.random.randn()
    dy0 = np.random.randn()

    # State has shape (time, variables)
    arrays[f"state{i}"] = np.zeros((N+1, 4))

    arrays[f"state{i}"][0] = np.array([
        x0,
        y0,
        dx0,
        dy0
    ])

    rk4(0, arrays[f"state{i}"], dt)


# Plot
plt.figure(figsize=(10, 5))

for i, arr in enumerate(arrays.values()):
    plt.plot(arr[:, 1], arr[:, 0], label=f"Run {i}")

plt.xlabel("y")
plt.ylabel("x")
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 5))

for i, arr in enumerate(arrays.values()):
    plt.plot(arr[:, 0], arr[:, 2], label=f"Run {i}")

plt.xlabel("x")
plt.ylabel("dx")
plt.legend()
plt.grid(True)
plt.show()

for i, arr in enumerate(arrays.values()):
    plt.plot(arr[:, 1], arr[:, 3], label=f"Run {i}")

plt.xlabel("y")
plt.ylabel("dy")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
def rk4(t, y, h1):                  # function for rk4 init
    for i in range(0, N):
        t  = h1 * i
        k0 = h1 * f( t, y[i] )
        k1 = h1 * f( t + h1/2., y[i] + k0/2. )
        k2 = h1 * f( t + h1/2., y[i] + k1/2. )
        k3 = h1 * f( t + h1, y[i] + k2 )
        y[i + 1] = y[i]  +  (1./6.) * (k0  +  2.*k1  +  2.*k2 + k3)
    return y[3]

def f(t, state):
    x, y, dx, dy = state

    f0 = dx / dy
    f1 = 1/m*F_x(dx, dy, dt, m, k, n)

    f2 = dy / dt
    f3 = 1/m*F_y(dx, dy, dt, m, k, n) - g

    return np.array([f0, f1, f2, f3])

def F_x(dx, dy, dt, m, k, n):
    dx_dt = dx / dt
    dy_dt = dy / dt
    F = m*(-k*dx_dt**n*dx_dt/(np.sqrt(dx_dt**2 + dy_dt**2)))
    return F
def F_y(dx, dy, dt, m, k, n):
    dx_dt = dx / dt
    dy_dt = dy / dt
    F = m*(-g-k*dy_dt**n*dy_dt/(np.sqrt(dx_dt**2 + dy_dt**2))) + m*g
    return F

# parameters
m = 1
x0 = 0
y0 = 0
dx0 = 1
dy0 = 1
dt = 0.1
N = 1000
k = 1
n = 1
g =9.8
arrays= np.zeros((N+1, 4))

arrays[0] = np.array([
        x0,
        y0,
        dx0,
        dy0
    ])

rk4(0, arrays, dt)

plt.plot(arrays[:, 0], arrays[:, 1])

plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
def rk4(t, y, h1):                  # function for rk4 init
    for i in range(0, N):
        t  = h1 * i
        k0 = h1 * f( t, y[i] )
        k1 = h1 * f( t + h1/2., y[i] + k0/2. )
        k2 = h1 * f( t + h1/2., y[i] + k1/2. )
        k3 = h1 * f( t + h1, y[i] + k2 )
        y[i + 1] = y[i]  +  (1./6.) * (k0  +  2.*k1  +  2.*k2 + k3)
    return y[3]

def f(dt, state):
    x, y, dx, dy = state

    f0 = -x*dt/r(x,y)**3
    f1 = -x/r(x,y)**3

    f2 = -y*dt/r(x,y)**3
    f3 = -y/r(x,y)**3

    return np.array([f0, f1, f2, f3])

def r(x,y):
    return np.sqrt(x**2 + y**2)
# parameters
dt = 0.0001
x0 = 0.5
y0 = 0
dx0 = 1
dy0 = 1.63*dt

N = 1000

arrays= np.zeros((N+1, 4))

arrays[0] = np.array([
        x0,
        y0,
        dx0,
        dy0
    ])

rk4(0, arrays, dt)
time_array = []

for i in range(N+1):
    time_array = np.append(time_array,dt*i)

plt.plot(time_array, arrays[:,0])

plt.xlabel("t")
plt.ylabel("x")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.integrate import solve_ivp
from scipy.optimize import brentq


# ============================================================
# Physical constants / units
# ============================================================

hbar = 1.0
m = 1.0



V0 = 10.0
a = 1.0


def V(x):
    """
    Finite square-well potential.
    """
    return np.where(np.abs(x) < a, -V0, 0.0)



def schrodinger_rhs(x, y, E):

    psi = y[0]
    dpsi = y[1]

    ddpsi = (
        2.0 * m / hbar**2
        * (V(x) - E)
        * psi
    )

    return np.array([
        dpsi,
        ddpsi
    ])


# ============================================================
# Numerical domain
# ============================================================

Xmax = 10.0

# Point at which the left and right solutions are matched
xmatch = 0.0




def mismatch(E, return_solutions=False):

    # --------------------------------------------------------
    # Bound states require E < 0 for this potential.
    # --------------------------------------------------------

    if E >= 0.0:
        raise ValueError(
            "E must be negative for a bound state."
        )



    kappa = np.sqrt(
        -2.0 * m * E
    ) / hbar


   

    psi_L_initial = 1.0
    dpsi_L_initial = kappa

    sol_L = solve_ivp(
        lambda x, y: schrodinger_rhs(x, y, E),


        (-Xmax, xmatch),


        [
            psi_L_initial,
            dpsi_L_initial
        ],

        rtol=1e-10,
        atol=1e-12,

        dense_output=True
    )

    if not sol_L.success:
        raise RuntimeError(
            "Left ODE integration failed:\n"
            + sol_L.message
        )

    psi_L = sol_L.y[0, -1]
    dpsi_L = sol_L.y[1, -1]




    psi_R_initial = 1.0
    dpsi_R_initial = -kappa

    sol_R = solve_ivp(
        lambda x, y: schrodinger_rhs(x, y, E),

        # Integrate from the right boundary toward the center
        (Xmax, xmatch),

        # Initial conditions
        [
            psi_R_initial,
            dpsi_R_initial
        ],

        # Numerical tolerances
        rtol=1e-10,
        atol=1e-12,


        dense_output=True
    )

    if not sol_R.success:
        raise RuntimeError(
            "Right ODE integration failed:\n"
            + sol_R.message
        )

    # Values at matching point
    psi_R = sol_R.y[0, -1]
    dpsi_R = sol_R.y[1, -1]




    W = (
        psi_L * dpsi_R
        - dpsi_L * psi_R
    )


    if return_solutions:
        return W, sol_L, sol_R

    return W




E_min = -V0 + 1e-8
E_max = -1e-8


# ============================================================
# Scan the energy range
# ============================================================

n_scan = 2000

energies = np.linspace(
    E_min,
    E_max,
    n_scan
)

wronskians = np.empty_like(energies)


for i, E in enumerate(energies):

    try:
        wronskians[i] = mismatch(E)

    except Exception:
        wronskians[i] = np.nan


# ============================================================
# Find intervals containing eigenvalues
#
# An eigenvalue corresponds to W(E) = 0.
#
# Therefore look for sign changes.
# ============================================================

root_intervals = []

for i in range(len(energies) - 1):

    E1 = energies[i]
    E2 = energies[i + 1]

    W1 = wronskians[i]
    W2 = wronskians[i + 1]

    if not np.isfinite(W1):
        continue

    if not np.isfinite(W2):
        continue

    if W1 == 0.0:

        root_intervals.append(
            (E1, E1)
        )

    elif W1 * W2 < 0.0:

        root_intervals.append(
            (E1, E2)
        )


# ============================================================
# Refine eigenvalues using Brent's method
# ============================================================

eigenvalues = []

for E1, E2 in root_intervals:

    # Exact grid point
    if E1 == E2:

        eigenvalues.append(E1)

        continue

    try:

        E_root = brentq(
            mismatch,
            E1,
            E2,

            # Absolute energy tolerance
            xtol=1e-12,

            # Relative tolerance
            rtol=1e-12
        )

        eigenvalues.append(E_root)

    except ValueError:
        pass


# ============================================================
# Remove duplicate roots
# ============================================================

if eigenvalues:

    eigenvalues = np.array(eigenvalues)

    eigenvalues = np.unique(
        np.round(
            eigenvalues,
            decimals=11
        )
    )

else:

    eigenvalues = np.array([])


# ============================================================
# Print eigenvalues
# ============================================================

print()
print("=" * 60)
print("Numerical bound-state energies")
print("=" * 60)

if len(eigenvalues) == 0:

    print("No eigenvalues found.")

else:

    for n, E in enumerate(eigenvalues):

        print(
            f"n = {n:2d}    "
            f"E = {E:.12f}"
        )

print("=" * 60)


# ============================================================
# Plot Wronskian
# ============================================================

plt.figure(figsize=(9, 5))

plt.plot(
    energies,
    wronskians,
    label=r"$W(E)$"
)

plt.axhline(
    0.0,
    linestyle="--",
    linewidth=1.0
)

for E in eigenvalues:

    plt.axvline(
        E,
        linestyle=":",
        linewidth=1.0
    )

plt.xlabel(r"Energy $E$")
plt.ylabel(r"Wronskian $W(E)$")

plt.title(
    "Eigenvalue Search Using Wronskian Matching"
)

plt.grid()
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# Plot the potential
# ============================================================

x_plot = np.linspace(
    -Xmax,
    Xmax,
    2000
)

plt.figure(figsize=(9, 5))

plt.plot(
    x_plot,
    V(x_plot),
    label=r"$V(x)$"
)

plt.xlabel("$x$")
plt.ylabel("$V(x)$")

plt.title("Finite Square-Well Potential")

plt.grid()
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# Construct and plot eigenfunctions
# ============================================================

for n, E in enumerate(eigenvalues):

    # --------------------------------------------------------
    # Recalculate the two solutions at this eigenvalue
    # --------------------------------------------------------

    W, sol_L, sol_R = mismatch(
        E,
        return_solutions=True
    )


    # --------------------------------------------------------
    # Create x grids
    # --------------------------------------------------------

    x_left = np.linspace(
        -Xmax,
        xmatch,
        1000
    )

    x_right = np.linspace(
        xmatch,
        Xmax,
        1000
    )


    # --------------------------------------------------------
    # Evaluate solutions
    # --------------------------------------------------------

    psi_left = sol_L.sol(x_left)[0]

    psi_right = sol_R.sol(x_right)[0]



    if abs(psi_right[0]) < 1e-14:

        print(
            f"Warning: right solution nearly zero "
            f"at matching point for n={n}"
        )

        continue

    scale = (
        psi_left[-1]
        / psi_right[0]
    )

    psi_right *= scale



    x = np.concatenate([
        x_left,
        x_right[1:]
    ])

    psi = np.concatenate([
        psi_left,
        psi_right[1:]
    ])



    norm = np.sqrt(
        np.trapezoid(
            np.abs(psi)**2,
            x
        )
    )

    psi /= norm




    center_index = np.argmin(
        np.abs(x - xmatch)
    )

    if psi[center_index] < 0:

        psi *= -1.0


    # --------------------------------------------------------
    # Plot eigenfunction
    # --------------------------------------------------------

    plt.figure(figsize=(9, 5))

    plt.plot(
        x,
        psi,
        label=fr"$\psi_{n}(x)$"
    )

    plt.axhline(
        0.0,
        linestyle="--",
        linewidth=1.0
    )


    V_plot = V(x)

    V_scale = np.max(
        np.abs(psi)
    )

    if np.max(np.abs(V_plot)) > 0:

        V_scaled = (
            V_plot
            / np.max(np.abs(V_plot))
            * V_scale
        )

        plt.plot(
            x,
            V_scaled,
            linestyle=":",
            label="scaled $V(x)$"
        )

    plt.xlabel("$x$")
    plt.ylabel(r"$\psi(x)$")

    plt.title(
        f"Eigenstate n = {n}, "
        f"E = {E:.8f}"
    )

    plt.grid()
    plt.legend()
    plt.tight_layout()
    plt.show()
```
