```python
# 14.3
import matplotlib.pyplot as plt

def function(f, i, dt, N_, N0):
    f = (2 * dt) / (1 + 2*dt * N_) * N0
    return f

dt = 0.01
N_ = 1000
N0 = 10
steps = 100000

x = []
T = []

for i in range(steps):
    x.append(function(x, i, dt, N_, N0))
    T.append(dt * i)

plt.plot(T, x)
plt.xlabel("Time")
plt.ylabel("x")
plt.show()
```
